```python
img = cv2.imread("H:\\HK1_2026-2027\\ComputerVision\\Code\\anhTest.jpg", cv2.IMREAD_GRAYSCALE)
img
```

```python
gamma = 2
img_gamma = ((img/255.0)**gamma) * 255
```

```python
img_gamma = np.clip(img_gamma, 0, 255).astype(np.uint8)
```
